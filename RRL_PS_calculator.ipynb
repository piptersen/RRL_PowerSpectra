```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import matplotlib.ticker as ticker

def plot_for_z(ax, z_plot, bw_plot='05', xlim=None, ylim=None):
    # --- Load data ---
    #k_perp = 0.0
    data1 = np.load(f'P_RRL_data/P_RRL_data_new/PS_z{z_plot}_bw{bw_plot}_kperp00_ne3_EM6.npz')
    k_1, P21_1, PRRL_1, P21RRL_1, PRRLRRL_1 = data1['k'], data1['P21'], data1['PRRL'], 2*data1['P21RRL'], data1['PRRLRRL']

    # k_perp = 0.1
    data2 = np.load(f'P_RRL_data/P_RRL_data_new/PS_z{z_plot}_bw{bw_plot}_kperp01_ne3_EM6.npz')
    k_2, P21_2, PRRL_2, P21RRL_2, PRRLRRL_2 = data2['k'], data2['P21'], data2['PRRL'], 2*data2['P21RRL'], data2['PRRLRRL']

    # --- Plot each component ---
    #ax.loglog(k_1, np.absolute(P21_1), c="#93A4C2", label="21cm auto (0.1)")
    #ax.loglog(k_1, np.absolute(PRRL_1), c="#C7A2A2", label="RRL auto (0.1)")
    #ax.loglog(k_1, np.absolute(P21RRL_1), c="#B4A067", label="21×RRL (0.1)")
    #ax.loglog(k_1, np.absolute(PRRLRRL_1), c="#537E53", label="RRL×RRL (0.1)")

    ax.loglog(k_1, np.absolute(P21_1), c="#2E538D", label="21cm auto (0.1)", lw = 2, alpha = 0.7)
    ax.loglog(k_1, np.absolute(PRRL_1), c="#9E0303", label="RRL auto (0.1)", lw = 2, alpha = 0.7)
    ax.loglog(k_1, np.absolute(P21RRL_1), c="#B8860B", label="21×RRL (0.1)", lw = 2, alpha = 0.7)
    ax.loglog(k_1, np.absolute(PRRLRRL_1), c="#013F01", label="RRL×RRL (0.1)", lw = 2, alpha = 0.7)

    ax.loglog(k_2, np.absolute(P21_2), ls="--", lw=2, c="#2E538D", label="21cm auto (0.0)")
    ax.loglog(k_2, np.absolute(PRRL_2), ls="--", lw=2, c="#9E0303", label="RRL auto (0.0)")
    ax.loglog(k_2, np.absolute(P21RRL_2), ls="--", lw=2, c="#B8860B", label="21×RRL (0.0)")
    ax.loglog(k_2, np.absolute(PRRLRRL_2), ls="--", lw=2, c="#013F01", label="RRL×RRL (0.0)")
    #ax.loglog(k_2, PRRLRRL_2, ls="--", lw=2, c="#644076", label="RRL×RRL (0.0)")

    if xlim: ax.set_xlim(*xlim)
    if ylim: ax.set_ylim(*ylim)

    # Redshift text inside panel
    if z_plot == '05': z_plot = '0.5'
    ax.text(0.05, 0.95, fr"$z_{{21}}={z_plot}$",
            transform=ax.transAxes, ha="left", va="top", fontsize=12)

    # Inward ticks on all sides
    ax.tick_params(axis="both", which="both", direction="in", top=True, right=True)

# --- Example usage ---
z_values = ['05', '1', '2', '5']
BW = '05' # Can be '025', '05', '1' for z = {0.025, 0.05, 0.1}
xlims = {'05': (5e-2, 3), '1': (5e-2, 3), '2': (5e-2, 3), '5': (5e-2, 3)}
ylims = {'05': (1e-8, 5e2), '1': (1e-8, 5e2), '2': (1e-6, 5e2), '5': (1e-6, 5e2)}

fig, axes = plt.subplots(2, 2, figsize=(8, 6), sharex=False, sharey=False)
for i, (ax, z) in enumerate(zip(axes.ravel(), z_values)):
    plot_for_z(ax, z, BW, xlim=xlims.get(z), ylim=ylims.get(z))

    # Remove redundant labels
    if i == 2:
        ax.set_yticks([1e-8, 1e-6, 1e-4, 1e-2, 1, 100])
        #ax.set_yticklabels([f"10^{tick:.0e}" for tick in ax.get_yticks()])
        ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: f"$10^{{{int(np.log10(y))}}}$"))
    if i < 2:  # top row
        ax.set_xlabel("")
        ax.set_xticklabels([])
    if i % 2 == 1:  # right col
        ax.set_ylabel("")
        ax.set_yticklabels([])
    
""
# Shared labels
fig.text(0.55, 0.04, r"$k\,[h\,\mathrm{Mpc}^{-1}]$", ha="center", fontsize = 12, color = 'black')
fig.text(0.04, 0.5, r"$\left|\Delta^2\right|(k)$", va="center", rotation="vertical", fontsize=12, color = 'black')

# Shared legend
solid_line   = mlines.Line2D([], [], color="black", linestyle="-",  label=r"$k_\perp=0.0$")
dashed_line  = mlines.Line2D([], [], color="black", linestyle="--", label=r"$k_\perp=0.1$")

cm21_line    = mlines.Line2D([], [], color="#2E538D", linestyle="-", label=r"$\Delta^2_{21}$")
rrl_line     = mlines.Line2D([], [], color="#9E0303", linestyle="-", label=r"$\Delta^2_{\mathrm{RRL}}$")
cross_line   = mlines.Line2D([], [], color="#B8860B", linestyle="-", label=r"$\Delta^2_{21\times \mathrm{RRL}}$")
rrlxrrl_line = mlines.Line2D([], [], color="#006400", linestyle="-", label=r"$\Delta^2_{\mathrm{RRL\times RRL}}$")


leg1 = fig.legend(handles=[solid_line, dashed_line],
           loc="upper center", ncol=2, frameon=True, fontsize=9, bbox_to_anchor=(0.55, 1))#, facecolor = 'black')
           
leg2 = fig.legend(handles=[cm21_line, rrl_line, cross_line, rrlxrrl_line],
              loc="upper center", ncol=4, frameon=True, fontsize=9, bbox_to_anchor=(0.55, 0.95))#, facecolor = 'black')

# Layout
#ax.add_artist(leg1)
plt.subplots_adjust(left=0.12, right=0.98, bottom=0.1, top=0.88,
                    wspace=0.05, hspace=0.05)
#plt.savefig("figures/PS_allterms_kperp.png", dpi=300)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import interpolate

# ------------------------
# User options
# ------------------------
transparent_mode = False    # Set True if you want transparent background
line_color = "white"        # Color of the matter PS line on dark background
# ------------------------

# Load data
file_path = 'initSB_matterpower_z0.dat'
h = 0.7

data = np.loadtxt(file_path)
k_data = data[:, 0] * h
P = data[:, 1] / h**3

# k-space
k_space = np.logspace(-1.5, 0.5, 500)

# Interpolate
model = interpolate.interp1d(k_data, P, fill_value="extrapolate")
matterPS = model(k_space)

# Convert to Δ²(k)
P_unitless = k_space**3 / (2 * np.pi**2) * matterPS

# ------------------------
# Plot setup
# ------------------------
fig, ax = plt.subplots(figsize=(6, 4))

# Backgrounds
if transparent_mode:
    fig.patch.set_alpha(0)
    ax.set_facecolor('none')
else:
    fig.patch.set_facecolor('black')
    ax.set_facecolor('black')

# Plot
ax.loglog(k_space, P_unitless, lw=2, color=line_color, label='Matter PS')
ax.loglog(k_space[50:225], P_unitless[50:225], lw=3, color="#E97132", label='BAO scales')

# Labels
ax.set_xlabel(r"$k\,[h\,\mathrm{Mpc}^{-1}]$", color='white', fontsize=12)
ax.set_ylabel(r"$\Delta^2(k)$", color='white', fontsize=12)

# Ticks + spines
ax.tick_params(axis='both', colors='white')
for spine in ax.spines.values():
    spine.set_color('white')

# Legend
leg = ax.legend(frameon=True)
leg.get_frame().set_facecolor('none' if transparent_mode else 'black')
leg.get_frame().set_edgecolor('white')
for txt in leg.get_texts():
    txt.set_color('white')

# Layout & save
plt.tight_layout()

# Save correctly
plt.savefig(
    "MatterPowerSpectrum.png",
    dpi=300,
    transparent=transparent_mode,
    facecolor=fig.get_facecolor()
)

plt.show()
```

```python
import numpy as np
from scipy.integrate import quad

def madauFit(z):
    return 0.015 * (1 + z)**2.7 / (1 + ((1 + z)/2.9)**5.6)

def sfr_schechter(psi, phi_star, psi_star, alpha):
    """
    dn/dpsi [Mpc^-3 (Msun/yr)^-1]
    """
    return (
        phi_star
        * (psi / psi_star)**alpha
        * np.exp(-psi / psi_star)
        / psi_star
    )

def A_prefactor(z, tau_L, N_HII, nu_emit, linewidth):
    nu_obs = nu_emit / (1 + z)

    A = (
        5.3e-5 / 0.1
        * (tau_L * N_HII)
        * (nu_obs / 150.0)**(-2.8)
        * (linewidth / 20.0)
        * ((1 + z) / 3.0)**(-2.3)
    )
    return A  # K per (Msun/yr)

def shot_noise_RRL(
    z,
    tau_L,
    N_HII,
    nu_emit,
    linewidth,
    phi_star,
    psi_star,
    alpha,
    psi_min=1e-3,
    psi_max=1e3,
):
    A = A_prefactor(z, tau_L, N_HII, nu_emit, linewidth)

    def integrand(psi):
        return sfr_schechter(psi, phi_star, psi_star, alpha) * psi**2

    I, _ = quad(integrand, psi_min, psi_max)

    P_shot = A**2 * I
    return P_shot  # [K^2 Mpc^3]


z = 1.0
tau_L = 1e-3
N_HII = 1
nu_emit = 100.0  # MHz
linewidth = 20.0

# Mock SFR function parameters
phi_star = 1e-3   # Mpc^-3
psi_star = 10.0   # Msun/yr
alpha = -1.6

Pshot = shot_noise_RRL(
    z, tau_L, N_HII, nu_emit, linewidth,
    phi_star, psi_star, alpha
)

print(f"Shot noise at z={z}: {Pshot:.3e} K^2 Mpc^3")
```

```python
import numpy as np
from scipy.integrate import quad

def n_eff_inverse_sfr(
    phi_star,
    psi_star,
    alpha,
    psi_min=1e-3,
    psi_max=1e3,
):
    """
    Effective inverse number density for shot noise.
    Returns n_eff^{-1} in units of Mpc^3.
    """

    def dn_dpsi(psi):
        return (
            phi_star
            * (psi / psi_star)**alpha
            * np.exp(-psi / psi_star)
            / psi_star
        )

    I1, _ = quad(lambda psi: psi * dn_dpsi(psi), psi_min, psi_max)
    I2, _ = quad(lambda psi: psi**2 * dn_dpsi(psi), psi_min, psi_max)

    return I2 / I1**2

phi_star = 1e-3   # Mpc^-3
psi_star = 10.0   # Msun/yr
alpha = -1.6

ninv = n_eff_inverse_sfr(phi_star, psi_star, alpha)
print(f"n_eff^-1 = {ninv:.3e} Mpc^3")
```

### Schechter
$\phi(\psi) = \frac{\phi_*}{\psi_*} ~ (\frac{\psi}{\psi_*})^\alpha ~ e^{(-\psi / \psi_*)}$

### Shot Noise
$P_{shot} = \int d\psi ~ \phi(\psi) T^2$

### $n_{eff}^{-1}$
$n_{eff}^{-1} = \frac{P_{shot}}{T^2}$

```python
import numpy as np
from scipy.integrate import quad
def sfr_schechter(psi, phi_star, psi_star, alpha):
    """
    dn/dpsi [Mpc^-3 (Msun/yr)^-1]
    """
    return phi_star * (psi / psi_star)**alpha * np.exp(-psi / psi_star) / psi_star

def freq_z(nu_emit, z):
    '''
    Calculates cosmological redshift of ground state H frequency
    '''
    return(nu_emit/(1+z))

def rest_freq(n):
    '''
    Calculates H transition frequency in MHz
    ------------------
    n            (int)   : Transition level
    ------------------
    return [MHz] (float) : Frequency of n transition
    '''
    return(((1 / (n)**2) - (1 / (n+1)**2))*3.288051231582647*1e15*1e-6)

def P_shot(z, phi_star, psi_star, alpha, T_line_func, n, psi_min, psi_max):
    """
    Computes shot noise for a given line temperature function
    """
    integrand = lambda psi: sfr_schechter(psi, phi_star, psi_star, alpha) * T_line_func(psi, z, n)**2
    Pshot, err = quad(integrand, psi_min, psi_max)
    return Pshot

def T_CO(psi, z, n, tau_L=1e-3, N_HII=100, linewidth=20.0):
    """
    Example temperature for CO(1-0) in K
    """
    nu_emit = rest_freq(n) * 1e-6
    E_SFR = psi #madauFit(z)
    nu_shift = freq_z(nu_emit, z)
    T_L = 5.3e-5 * (E_SFR / 0.1) * ((tau_L * N_HII) / 0.1) * (nu_shift / 150)**(-2.8) * (linewidth / 20) * ((1+z)/3)**(-2.3)
    return(T_L)

def T_mean(z, phi_star, psi_star, alpha, T_line_func, n,
           psi_min, psi_max):

    integrand = lambda psi: sfr_schechter(
        psi, phi_star, psi_star, alpha
    ) * T_line_func(psi, z, n)

    Tbar, err = quad(integrand, psi_min, psi_max)
    return Tbar

def n_eff_inv(z, phi_star, psi_star, alpha, T_line_func, n,
              psi_min, psi_max):

    Tbar = T_mean(z, phi_star, psi_star, alpha,
                  T_line_func, n,
                  psi_min, psi_max)

    Pshot = P_shot(z, phi_star, psi_star, alpha,
                   T_line_func, n,
                   psi_min, psi_max)

    return Pshot / Tbar**2

z = 3
#check 
phi_star = 1e-3
psi_star = 200

psi_min = 1e-5
psi_max = 1000

alpha = -1.6
n = 166

n_eff_inv_CO = n_eff_inv(
    z, phi_star, psi_star, alpha, T_CO, n, psi_min, psi_max
)

print(f"n_eff^-1 at z={z}: {n_eff_inv_CO:.3e} Mpc^3")
```
