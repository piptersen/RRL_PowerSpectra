```python
import numpy as np
import matplotlib.pyplot as plt


# -----------------------------
# settings
# -----------------------------
DATA_DIR = "P_RRL_data"
BAND_STR = "05"                      # "05" in filenames
Z_PANELS = [0.5, 1, 2, 5]            # 4 stacked panels

# 4 models total: your 3 + the tau01 variant (unique model)
MODELS = [
    {"name": r"Model A",         "ne": "ne1", "EM": "EM4", "tau01": False},
    {"name": r"Model B",         "ne": "ne3", "EM": "EM5", "tau01": False},
    {"name": r"Model C",         "ne": "ne3", "EM": "EM6", "tau01": False},
    {"name": r"Model D",         "ne": "ne4", "EM": "EM7", "tau01": False},
    {"name": r"PO11",         "ne": "ne3", "EM": "EM6", "tau01": True},
    
]


def z_to_str(z):
    """Match your filename convention (z=0.5 is '05')."""
    if z == 0.5:
        return "05"
    if z == 1:
        return "1"
    if z == 2:
        return "2"
    if z == 5:
        return "5"
    return str(z)


# -----------------------------
# helpers (close to your original math)
# -----------------------------
def smooth_logpoly(k, P, kmin, kmax, poly_order=3):
    mask = (k > kmin) & (k < kmax) & (P > 0)
    k_fit = k[mask]
    P_fit = P[mask]
    coeffs = np.polyfit(np.log(k_fit), np.log(P_fit), poly_order)
    return np.exp(np.polyval(coeffs, np.log(k)))


def find_k_roots_from_sign_change(k, y):
    sign_change = np.where(np.diff(np.sign(y)))[0]
    k_roots = []
    for i in sign_change:
        k0, k1 = k[i], k[i + 1]
        y0, y1 = y[i], y[i + 1]
        k_roots.append(k0 - y0 * (k1 - k0) / (y1 - y0))
    return np.array(k_roots)


def compute_frac_shift(k, P21, PRRL, P21RRL, PRRLRRL):
    P_smooth_21 = smooth_logpoly(k, P21, kmin=1e-2, kmax=10, poly_order=3)

    P_contaminated = np.abs(PRRL) + np.abs(P21RRL) + np.abs(PRRLRRL)
    P_smooth_rrl = smooth_logpoly(k, P_contaminated, kmin=1e-5, kmax=1, poly_order=3)

    norm = 2 * np.pi**2 / k**3
    P21_normalized = (P21 - P_smooth_21) * norm
    PRRL_normalized = (P_contaminated - P_smooth_rrl) * norm

    dPdk = np.gradient(P21_normalized, k)
    d2Pdk2 = np.gradient(dPdk, k)
    dPRRLdk = np.gradient(PRRL_normalized, k)

    k_roots = find_k_roots_from_sign_change(k, dPdk)

    dPRRL_interp = np.interp(k_roots, k, dPRRLdk)
    d2P_interp = np.interp(k_roots, k, d2Pdk2)

    delta_k = -dPRRL_interp / d2P_interp
    frac_shift = delta_k / k_roots
    return k_roots, frac_shift


def filename_for_model(z_str, band_str, model, kperp="00"):
    base = f"{DATA_DIR}/PS_z{z_str}_bw{band_str}_kperp{kperp}_{model['ne']}_{model['EM']}"
    if model["tau01"]:
        return base + "_tau01.npz"
    return base + ".npz"
```

```python


# -----------------------------
# plot (4 stacked, shared x, no vertical gaps, legend on bottom)
# -----------------------------
fig, axes = plt.subplots(
    nrows=4, ncols=1, sharex=True, figsize=(7, 10))#,
    #gridspec_kw={"hspace": 0.0}  # no gap between panels
#)

colors = ['crimson', "saddlebrown", '#177E89', 'goldenrod', 'black']
markers = ["s", "x", "^", "o", "s"]
#colors = ["#2a96cc", "#f4d90f", "#cc2a96", "#96cc2a", 'black']
sis = [36, 26, 26, 26, 26]

for ax, z in zip(axes, Z_PANELS):
    z_str = z_to_str(z)

    for model, marker, color, spwah in zip(MODELS, markers, colors, sis):
        fn = filename_for_model(z_str, BAND_STR, model)
        dat = np.load(fn)

        k = dat["k"]
        k_roots, frac_shift = compute_frac_shift(
            k, dat["P21"], dat["PRRL"], 2*dat["P21RRL"], dat["PRRLRRL"]
        )
        if model["tau01"] == False:
            ax.scatter(k_roots, frac_shift, s=spwah, marker=marker, label=model["name"], color=color)
            ax.plot(k_roots, frac_shift, lw=1, color=color, alpha = 0.6, ls = '--')
        if model["tau01"]:
            print(f"tau=0.1 model: mean abs frac shift = {np.nanmean(np.abs(frac_shift[:18])):.2e}")

    ax.axhline(0, color="k", ls="--", lw=1)
    ax.grid(True)
    ax.set_ylabel(r"$\Delta k/k$")
    ax.text(
        0.98, 0.90, rf"$\bf{{z_{{21}}={z:g}}}$",
        transform=ax.transAxes, ha="right", va="top", fontsize = 12, color="black", backgroundcolor="white", weight = "bold"
    )
    ax.set_yticks([-8e-5, -4e-5, 0,  4e-5, 8e-5])

axes[-1].set_xlabel(r"$k\ [h/\mathrm{Mpc}]$")
axes[-1].set_xlim(0.05, 0.30)
for ax in axes:
    ax.set_ylim(-10e-5, 10e-5)

handles, labels = axes[-1].get_legend_handles_labels()
#axes[-1].legend(handles, labels, loc="lower right", frameon=True)
axes[-1].set_ylim(-10e-5, 10e-5)
handles, labels = axes[-1].get_legend_handles_labels()
fig.legend(handles[:], labels[:],
           loc="upper center",
           bbox_to_anchor=(0.54, 0.99),
           ncol=4,
           frameon=True, fontsize=10)

# leave space for the legend
fig.tight_layout(rect=[0, 0, 1, 0.97])

#plt.savefig('figures/RRL_cosmology_calculator_modelA.png', dpi=300)
```

```python
# ---------------------------------------------------------
# Median absolute BAO-peak shift versus redshift
# Points: median |Delta k / k| for peaks with 0.1 < k < 1
# Error bars: full minimum-to-maximum range
# ---------------------------------------------------------

fig, ax = plt.subplots(figsize=(7, 5.5))

colors = ['crimson', 'saddlebrown', '#177E89', 'goldenrod', 'black']
markers = ['s', 'x', '^', 'o', 's']

plot_models = [
    (model, marker, color)
    for model, marker, color in zip(MODELS, markers, colors)
    if not model["tau01"]
]

z_values = np.sort(np.asarray(Z_PANELS, dtype=float))

# Horizontal separation between models, in redshift units
offset_step = 0.05

for model_index, (model, marker, color) in enumerate(plot_models):
    # Center all model offsets around the true redshift
    z_offset = (
        model_index - (len(plot_models) - 1) / 2
    ) * offset_step

    shifted_z = z_values + z_offset

    # ...the rest of the calculations remain unchanged...
    median_shifts = []
    lower_errors = []
    upper_errors = []

    for z in z_values:
        z_str = z_to_str(z)

        fn = filename_for_model(z_str, BAND_STR, model)
        dat = np.load(fn)

        k_roots, frac_shift = compute_frac_shift(
            dat["k"],
            dat["P21"],
            dat["PRRL"],
            2 * dat["P21RRL"],
            dat["PRRLRRL"]
        )

        k_roots = np.asarray(k_roots, dtype=float)
        frac_shift = np.asarray(frac_shift, dtype=float)

        # Keep finite BAO peaks within the requested k range
        mask = (
            (k_roots > 0.05)
            & (k_roots < 0.3)
            & np.isfinite(k_roots)
            & np.isfinite(frac_shift)
        )

        # Use magnitudes so positive and negative shifts do not cancel
        shifts = np.abs(frac_shift[mask])

        if len(shifts) == 0:
            median_shifts.append(np.nan)
            lower_errors.append(np.nan)
            upper_errors.append(np.nan)
            continue

        median_shift = np.median(shifts)
        minimum_shift = np.min(shifts)
        maximum_shift = np.max(shifts)

        median_shifts.append(median_shift)

        # Error-bar lengths measured from the median
        lower_errors.append(median_shift - minimum_shift)
        upper_errors.append(maximum_shift - median_shift)

    median_shifts = np.asarray(median_shifts)
    lower_errors = np.asarray(lower_errors)
    upper_errors = np.asarray(upper_errors)

    yerr = np.vstack([
        lower_errors,
        upper_errors
    ])

    ax.errorbar(
        shifted_z,
        median_shifts,
        yerr=yerr,
        marker=marker,
        markersize=6,
        color=color,
        linestyle="--",
        linewidth=1,
        elinewidth=1.3,
        capsize=6,
        capthick=2.0,
        label=model["name"]
    )

ax.set_xlabel(r"$z_{21}$")
ax.set_ylabel(r"Median absolute BAO peak shift, $|\Delta k/k|$")

ax.set_xticks(z_values)
ax.grid(True, alpha=0.35)

ax.legend(
    loc="best",
    ncol=2,
    frameon=True,
    fontsize=10
)

fig.tight_layout()
plt.semilogy()

# plt.savefig(
#     "figures/RRL_median_BAO_shift_vs_redshift.png",
#     dpi=300,
#     bbox_inches="tight"
# )
```
